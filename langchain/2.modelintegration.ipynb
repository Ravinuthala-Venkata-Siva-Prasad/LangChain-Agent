```python
import os
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-120b")
response = model.invoke("Why do parrots talk?")
response
```

```python
for chunk in model.stream(
    "Write about Artificial Intelligence in 3 separate paragraphs, "
    "with each paragraph around 60-70 words."
):
    print(chunk.text, end="", flush=True)
```

```python
model.invoke("Write about Artificial Intelligence in 3 separate paragraphs, "
    "with each paragraph around 60-70 words")
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="", flush=True)
```

```python
responses = model.batch([
    "what is AI?",
    "What is langchain?",
    "what is update are in langchain?"
])
for response in responses:
    print(response)
```
